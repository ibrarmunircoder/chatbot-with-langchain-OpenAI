### String Output Parser

```python
from dotenv import load_dotenv
import os

# Don't need to specify full path if notbook file and .env file are on the same leve in the project structure
load_dotenv("../.env")
```

```python
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.messages import HumanMessage
from langchain_core.output_parsers import StrOutputParser
```

```python
chat = ChatOpenAI(
    model="gpt-5",
    seed=365, 
    temperature=0
)
```

```python
message_h = HumanMessage(
    content="""Can you give me interesting fact that I probably didn't know about?"""
)
```

```python
response = chat.invoke([message_h])
```

```python
response
```

```python
str_output_parser = StrOutputParser()
```

```python
response_parsed = str_output_parser.invoke(response)
```

```python
response_parsed
```

### Comma-Seperated List Output Parser

```python
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.messages import HumanMessage
from langchain_core.output_parsers import CommaSeparatedListOutputParser
```

```python
chat = ChatOpenAI(
    model="gpt-5",
    seed=365, 
    temperature=0
)
```

```python
message_h = HumanMessage(
    content="""I've recently adopted a dog. Could you suggest some dog names?"""
)
```

```python
response = chat.invoke([message_h])
```

```python
print(response)
```

```python
list_output_parser = CommaSeparatedListOutputParser()
```

```python
response_parsed = list_output_parser.invoke(response)
```

How can we arrive at the desired result?

By instructing the chat model!

![list](../images/image-4.png)

```python
message_h = HumanMessage(
    content=f"""I've recently adopted a dog. Could you suggest some dog names?
    {CommaSeparatedListOutputParser().get_format_instructions()}
    """
)
message_h
```

```python
response = chat.invoke([message_h])
```

```python
response
```

```python
list_output_parser = CommaSeparatedListOutputParser()
```

```python
response_parsed = list_output_parser.invoke(response)
```

### Datetime Output Parser

```python
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.messages import HumanMessage
from langchain_classic.output_parsers.datetime import DatetimeOutputParser
chat = ChatOpenAI(
    model="gpt-5",
    seed=365, 
    temperature=0
)
```

```python
message_h = HumanMessage(
    content=f"""When was the Danish poet Hein born?
    {DatetimeOutputParser().get_format_instructions()}
    """
)
message_h
```

```python
response = chat.invoke([message_h])
```

```python
date_output_parser = DatetimeOutputParser()
```

```python
response_parsed = date_output_parser.invoke(response)
```
