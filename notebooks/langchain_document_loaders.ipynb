### Indexing: Text Embedding with OpenAI

```python
from dotenv import load_dotenv
import os

# Don't need to specify full path if notbook file and .env file are on the same leve in the project structure
load_dotenv("../.env")
```

```python
from langchain_community.document_loaders import Docx2txtLoader
from langchain_text_splitters.markdown import MarkdownHeaderTextSplitter
from langchain_text_splitters.character import CharacterTextSplitter
from langchain_openai.embeddings import OpenAIEmbeddings
import numpy as np
```

```python
loader_docx = Docx2txtLoader("Introduction_to_Data_and_Data_Science_2.docx")
pages = loader_docx.load()

md_splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on = [("#", "Course Title"),
                           ("##", "Lecture Title")]
)

pages_md_split = md_splitter.split_text(pages[0].page_content)

for i in range(len(pages_md_split)):
    pages_md_split[i].page_content = ' '.join(pages_md_split[i].page_content.split())

char_splitter = CharacterTextSplitter(
    separator = ".",
    chunk_size = 500,
    chunk_overlap = 50
)

pages_char_split = char_splitter.split_documents(pages_md_split)
```

```python
pages_char_split
```

```python
embedding = OpenAIEmbeddings(
    model="text-embedding-ada-002"
)
```

```python
vector1 = embedding.embed_query(pages_char_split[3].page_content)
vector2 = embedding.embed_query(pages_char_split[5].page_content)
vector3 = embedding.embed_query(pages_char_split[18].page_content)
len(vector1) # 1536
```

The embedding model we are using produces vectors with a dimensionality of 1536. The lengths of all the vectors are
the same. Therefore, we can apply the dot product to each pair of vectors and quantify their similarity. 

![vector](../images/image-19.png)

```python
np.dot(vector1, vector2), np.dot(vector1, vector3), np.dot(vector2, vector3), 
```

![magnitude](../images/image-20.png)

```python
np.linalg.norm(vector1), np.linalg.norm(vector2), np.linalg.norm(vector3)
```

```python
from dotenv import load_dotenv
import os

# Don't need to specify full path if notbook file and .env file are on the same leve in the project structure
load_dotenv("../.env")
```

```python
from langchain_community.document_loaders import Docx2txtLoader
from langchain_core.documents import Document
from langchain_text_splitters.markdown import MarkdownHeaderTextSplitter
from langchain_text_splitters.character import CharacterTextSplitter
from langchain_openai.embeddings import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
```

```python
loader_docx = Docx2txtLoader("Introduction_to_Data_and_Data_Science_2.docx")
pages = loader_docx.load()

md_splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on = [("#", "Course Title"),
                           ("##", "Lecture Title")]
)

pages_md_split = md_splitter.split_text(pages[0].page_content)

for i in range(len(pages_md_split)):
    pages_md_split[i].page_content = ' '.join(pages_md_split[i].page_content.split())

char_splitter = CharacterTextSplitter(
    separator = ".",
    chunk_size = 500,
    chunk_overlap = 50
)

pages_char_split = char_splitter.split_documents(pages_md_split)

embedding = OpenAIEmbeddings(
    model="text-embedding-ada-002"
)
```

### Retrieval: Similarity Search

```python
from langchain_core.documents import Document
from langchain_openai.embeddings import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
```

```python
embedding = OpenAIEmbeddings(
    model="text-embedding-ada-002"
)
```

```python
vectorstore_from_directory = Chroma(persist_directory="./intro-to-ds-lectures", 
                                    embedding_function=embedding)
```

```python
added_document = Document(
    page_content='Alright! So… Let’s discuss the not-so-obvious differences between the terms analysis and analytics. Due to the similarity of the words, some people believe they share the same meaning, and thus use them interchangeably. Technically, this isn’t correct. There is, in fact, a distinct difference between the two. And the reason for one often being used instead of the other is the lack of a transparent understanding of both. So, let’s clear this up, shall we? First, we will start with analysis',
    metadata={'Course Title': 'Intro to Data and Data Science',
              'Lecture Title': 'Analysis and Analytics'}
)
```

```python
vectorstore_from_directory.add_documents([added_document])
```

![Goal](../images/image-25.png)

```python
question = "What programming languages do data science use?"
```

```python
# k = 5 number of documents
retrieved_dcos = vectorstore_from_directory.similarity_search(query=question,
                                                              k=5)
```

```python
for i in retrieved_dcos:
    print(f"Page Content: {i.page_content}\n----------\nLecture Title:{i.metadata['Lecture Title']}\n")
```

### Maximum Marginal Relevance Search

```python
# k = 5 number of documents
retrieved_dcos = vectorstore_from_directory.max_marginal_relevance_search(
    query=question, 
    k=5,
    lambda_mult=0.1,
)
```

```python
for i in retrieved_dcos:
    print(f"Page Content: {i.page_content}\n----------\nLecture Title:{i.metadata['Lecture Title']}\n")
```

Now is a good time to introduce another parameter available for both search methods. Filtet instructs the method
to retrieve chunks only from the second lecture in the docx file.

```python
# k = 5 number of documents
retrieved_dcos = vectorstore_from_directory.max_marginal_relevance_search(
    query=question, 
    k=5,
    lambda_mult=0.1,
    filter={"Lecture Title": "Programming Language & Software Employed in Data Science - All the Tools You Need."}
)
```

### Generation: Stuffing Documents

```python
from langchain_openai.embeddings import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
from langchain_core.runnables import RunnablePassthrough
from langchain_core.runnables import RunnableParallel
from langchain_core.output_parsers import StrOutputParser
```

```python
vectorstore = Chroma(persist_directory="./intro-to-ds-lectures", 
                    embedding_function=OpenAIEmbeddings(model="text-embedding-ada-002")
                    )
```

```python
vectorstore.get()
```

```python
retriever = vectorstore.as_retriever(
    search_type="mmr",
    search_kwargs={"k": 3, "lambda_mult": 0.7},
)
```

```python
TEMPLATE = '''
Answer the following question:
{question}

To answer the question, use only the following context:
{context}

At the end of the response, specify the name of the lecture this context is taken from in the format:
Resources: *Lecture Title*
where *Lecture Title* should be substituted with the title of all resource lectures.
'''

prompt_template = PromptTemplate.from_template(TEMPLATE)
```

```python
chat = ChatOpenAI(
    model="gpt-5",
    seed=365, 
    temperature=0
)
```

```python
question = "What software do data science use?"
```

```python
chain = {
    "context": retriever,
    "question": RunnablePassthrough()
} | prompt_template | chat | StrOutputParser()
```

```python
chain.invoke(question)
```
