```python
import openai

client = openai.OpenAI()
```

```python
# Define a completion variable that will serve as our chatbot object

completion = client.chat.completions.create(
    model="gpt-4",
    # These messages will be a way to instruct the model on how to behave. It expects a list of dicts each
    # containing two required key value pairs
    messages=[
        {'role': 'system', 'content': """You are Marv, a chatbot that reluctantly answers questions with
        sarcastic responses."""},
        {'role': 'user', 'content': '''I've recently adopted a dog. Could you sugges some dog names?'''},
    ]
)
```

```python
# You will see that completion is an instance of ChatCompletion class
completion
```

```python
print(completion.choices[0].message.content)
```

### Temperature, Max Tokens, and Streaming

```python
# Define a completion variable that will serve as our chatbot object

completion = client.chat.completions.create(
    model="gpt-4",
    # These messages will be a way to instruct the model on how to behave. It expects a list of dicts each
    # containing two required key value pairs
    messages=[
        {'role': 'system', 'content': '''You are Marv, a chatbot that reluctantly answers questions with
        sarcastic responses. '''},
        {'role': 'user', 'content': '''Could you briefly explain what a block hole is?'''},
    ],
    max_tokens=150,
    temperature=0
    stream=True
)
```

```python
completion
```

```python
for i in completion:
    print(i.choices[0].delta.content, end ="")
```
